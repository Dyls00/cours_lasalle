````python
# Appliquez vos stratégies de traitement des valeurs manquantes
#Appliquer la moyenne sur les valeurs manques de tous les autres champs
df_n['sale_price'] = df_n['sale_price'].fillna(df['sale_price'].mean())
df_n['age'] = df_n['age'].fillna(df['age'].mean())
````

````python
#Colonnes supprimer
df_n = df_n.drop(columns=['notes', 'returned_at'])
````

````python
#convertir les dates en datetime
df_n['shipped_at'] = df_n['shipped_at'] = pd.to_datetime(df_n['shipped_at'])
df_n['delivered_at'] = df_n['delivered_at'] = pd.to_datetime(df_n['delivered_at'])
````

````python
#remplacer les champs vide des dates
df_n['shipped_at'] = df_n['shipped_at'].fillna(df_n['shipped_at'].median())
df_n['delivered_at'] = df_n['delivered_at'].fillna(df_n['delivered_at'].median())
````

````python
#remplacer les champs vide de category
df_n['category'] = df_n['category'].fillna(df_n['category'].mode()[0])
````

````python
#verifier mes données
df_n.isna().sum()
````

````python
# Détectez les doublons exacts et proposez une solution
df_n.duplicated
````

````python
#supprimer les doublons exacts
df_n = df_n.drop_duplicates()
````
